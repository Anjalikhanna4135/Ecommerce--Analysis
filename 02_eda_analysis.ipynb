```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/cleaned_data.csv")
```

```python
total_revenue = df['revenue'].sum()
total_orders = df['order_id'].nunique()
total_customers = df['customer_id'].nunique()

print(total_revenue, total_orders, total_customers)
```

```python
total_revenue = df['revenue'].sum()
total_orders = df['order_id'].nunique()
total_customers = df['customer_id'].nunique()

print(total_revenue, total_orders, total_customers)
```

```python
monthly_sales = df.groupby(['year','month'])['revenue'].sum().reset_index()

plt.figure()
sns.lineplot(data=monthly_sales, x='month', y='revenue')
plt.title("Monthly Revenue Trend")
plt.show()
```

```python
state_sales = df.groupby('customer_state')['revenue'].sum().sort_values(ascending=False).head(10)

plt.figure()
state_sales.plot(kind='bar')
plt.title("Top States by Revenue")
plt.show()
```

```python
customer_sales = df.groupby('customer_id')['revenue'].sum().sort_values(ascending=False)

top_10 = customer_sales.head(10)
```

```python
top_20_percent = int(len(customer_sales) * 0.2)
top_revenue = customer_sales.head(top_20_percent).sum()
total_revenue = customer_sales.sum()

print(top_revenue / total_revenue)
```

```python
orders_trend = df.groupby(['year','month'])['order_id'].nunique().reset_index()

plt.figure()
sns.lineplot(data=orders_trend, x='month', y='order_id')
plt.title("Order Volume Trend")
plt.show()
```

```python
orders_trend = df.groupby(['year','month'])['order_id'].nunique().reset_index()

plt.figure()
sns.lineplot(data=orders_trend, x='month', y='order_id')
plt.title("Order Volume Trend")
plt.show()
```

```python
payment_type = df['payment_type'].value_counts()

plt.figure()
payment_type.plot(kind='bar')
plt.title("Payment Method Distribution")
plt.show()
```
